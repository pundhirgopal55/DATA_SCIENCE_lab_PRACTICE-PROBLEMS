Q1. Create a CSV file of 10 students and read it using pd.read_csv(). Display the first 5 rows and info() summary.

Q2. Read the same file with header=None and names=[...]. Then use index_col to make Roll No the index.

```python
import pandas as pd

df = pd.read_csv(
    'students.csv',
    header=None,
    names=['Roll No', 'Name', 'Branch', 'Marks'],
    skiprows=1
)

# Make Roll No the index
df = pd.read_csv(
    'students.csv',
    header=0,
    names=['Roll No', 'Name', 'Branch', 'Marks'],
    index_col='Roll No'
)

print(df)
```

Q3. The CSV has missing values written as NA, - and blank cells. Load it so all three are read as NaN, then count missing values per column.

```python
import pandas as pd

data = {
    'Roll No': [101, 102, 103, 104, 105],
    'Name': ['Amit', 'Rahul', 'NA', 'Neha', 'Rohit'],
    'Branch': ['CSE', '-', 'IT', 'ECE', ''],
    'Marks': [85, 90, '-', 88, 'NA']
}

df = pd.DataFrame(data)

# Create CSV file
df.to_csv('missing_students.csv', index=False)

# Read CSV with missing values
df = pd.read_csv(
    'missing_students.csv',
    na_values=['NA', '-', '']
)

# Count missing values
print(df.isnull().sum())
```

Q6. Load a JSON file with json.loads() and convert it to a DataFrame. Then export the DataFrame back to JSON using to_json(orient="records").

```python
import pandas as pd
import json

# JSON data
data = '''
[
    {"Roll No": 101, "Name": "Amit", "Marks": 85},
    {"Roll No": 102, "Name": "Rahul", "Marks": 90},
    {"Roll No": 103, "Name": "Priya", "Marks": 78},
    {"Roll No": 104, "Name": "Neha", "Marks": 88}
]
'''

# Save JSON data into a file
with open('students.json', 'w') as f:
    f.write(data)

# Read JSON file
with open('students.json', 'r') as f:
    json_data = json.loads(f.read())

# Convert JSON into DataFrame
df = pd.DataFrame(json_data)

print(df)

# Convert DataFrame back to JSON
result = df.to_json(orient='records')

print(result)

# Save JSON output
with open('output.json', 'w') as f:
    f.write(result)
```

Q7. Read a large CSV in chunks of 1000 rows (chunksize) and find the average of one column across all chunks.

```python
import pandas as pd

total = 0
count = 0

# Read CSV in chunks
for chunk in pd.read_csv('students.csv', chunksize=1000):

    total += chunk['Marks'].sum()
    count += chunk['Marks'].count()

# Calculate average
average = total / count

print("Average Marks:", average)
```

Q8. Read two sheets from an Excel file using pd.ExcelFile and pd.read_excel(). Write a DataFrame to Excel with to_excel().

```python
import pandas as pd

# Create sample data
data1 = {
    'Roll No': [101, 102, 103],
    'Name': ['Amit', 'Rahul', 'Priya'],
    'Marks': [85, 90, 78]
}

data2 = {
    'Roll No': [104, 105, 106],
    'Name': ['Neha', 'Rohit', 'Anjali'],
    'Marks': [88, 76, 95]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

# Create Excel file with two sheets
with pd.ExcelWriter('students.xlsx') as writer:
    df1.to_excel(writer, sheet_name='Sheet1', index=False)
    df2.to_excel(writer, sheet_name='Sheet2', index=False)

# Read Excel file
excel = pd.ExcelFile('students.xlsx')

# Display sheet names
print(excel.sheet_names)

# Read first sheet
sheet1 = pd.read_excel(excel, sheet_name='Sheet1')

# Read second sheet
sheet2 = pd.read_excel(excel, sheet_name='Sheet2')

print("First Sheet:")
print(sheet1)

print("Second Sheet:")
print(sheet2)

# Export DataFrame to Excel
sheet1.to_excel('output.xlsx', index=False)
```

Q9. Use pd.read_html() on a Wikipedia page containing a table. Find how many tables were returned and display the required table.

```python
import pandas as pd
import requests
import io

# Wikipedia URL
url = 'https://en.wikipedia.org/wiki/List_of_countries_by_population_(United_Nations)'

# Set a User-Agent header to avoid 403 Forbidden error
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# Fetch the page content
response = requests.get(url, headers=headers)

# Wrap raw HTML in StringIO to avoid the deprecation warning
html_io = io.StringIO(response.text)

# Read HTML tables
tables = pd.read_html(html_io)

# Count total tables
print("Total tables:", len(tables))

# Display first table
print(tables[0])
```

Q10. Save a DataFrame using to_pickle() and load it back using read_pickle(). Compare CSV and Pickle file sizes.

```python
import pandas as pd
import os

# Create DataFrame
data = {
    'Roll No': [101, 102, 103, 104, 105],
    'Name': ['Amit', 'Rahul', 'Priya', 'Neha', 'Rohit'],
    'Marks': [85, 90, 78, 88, 76]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv('students.csv', index=False)

# Save as Pickle
df.to_pickle('students.pkl')

# Load Pickle file
df2 = pd.read_pickle('students.pkl')

print("Loaded DataFrame:")
print(df2)

# Compare file sizes
csv_size = os.path.getsize('students.csv')
pickle_size = os.path.getsize('students.pkl')

print("CSV file size:", csv_size, "bytes")
print("Pickle file size:", pickle_size, "bytes")
```
